### 🔐 Step 1: Secure Configuration
Enter your SendGrid credentials below to enable Email notifications and Impact Reports.

```python
from getpass import getpass

print("🔐 ENTERPRISE AUTHENTICATION (SENDGRID)")
sendgrid_key = getpass("Enter SENDGRID_API_KEY: ")
sendgrid_from = input("Enter SENDGRID_FROM_EMAIL (Must be verified in SendGrid): ")
gov_email = input("Enter GOVERNANCE_RECIPIENT_EMAIL: ")

with open(".env", "w") as f:
    f.write(f"SENDGRID_API_KEY={sendgrid_key}\n")
    f.write(f"SENDGRID_FROM_EMAIL={sendgrid_from}\n")
    f.write(f"GOVERNANCE_RECIPIENT={gov_email}\n")
    f.write("ENTERPRISE_MODE=DEMO\n")
    f.write("EMAIL_PROVIDER=SENDGRID\n")

print("\n[SUCCESS] Credentials securely stored in .env")
```
